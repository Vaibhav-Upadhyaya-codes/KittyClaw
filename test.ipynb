```python
import sys
import time
import itertools


def thinking_plus(duration=8):
    """
    Small stylish dancing plus animation
    """

    frames = itertools.cycle([
        "✚",
        "✖",
        "✚",
        "✜",
        "✚",
        "✛",
        "✚",
        "✢"
    ])

    start = time.time()

    while time.time() - start < duration:
        sys.stdout.write(f"\rThinking {next(frames)}")
        sys.stdout.flush()
        time.sleep(0.12)

    sys.stdout.write("\r" + " " * 30 + "\r")


thinking_plus()
```

```python
import sys
import time
import itertools

# ANSI TrueColor
ACCENT = "\033[38;2;34;211;238m"   # #22D3EE
TEXT   = "\033[38;2;230;247;255m"  # #E6F7FF
BG     = "\033[48;2;11;16;32m"     # #0B1020
RESET  = "\033[0m"


def thinking_plus(duration=8):
    """
    Cyan themed Thinking + animation
    Only 'Thinking' and symbol are colored
    """

    frames = itertools.cycle([
        "✚",
        "✢",
        "✣",
        "✤",
        "✥",
        "✦",
        "✧",
        "✜",
        "✛",
        "✖",
        "✚"
    ])

    start = time.time()

    while time.time() - start < duration:
        symbol = next(frames)

        sys.stdout.write(
            f"\r{BG}{ACCENT}Thinking {symbol}{RESET}"
        )
        sys.stdout.flush()
        time.sleep(0.11)

    sys.stdout.write("\r" + " " * 40 + "\r")


thinking_plus()
```
